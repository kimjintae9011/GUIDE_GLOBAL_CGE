```python
import pandas as pd
import numpy as np
from openpyxl import Workbook
from openpyxl.utils.dataframe import dataframe_to_rows
```

```python
WORLDBIG1 = pd.read_csv("WORLDBIG1.TXT", sep="\s+", header=None)
WORLDBIG2 = pd.read_csv("WORLDBIG2.TXT", sep="\s+", header=None)
```

```python
#raw data combine
WORLDBIG1.columns = ["region", "commodity", "year", "type", "unit", "value"]
WORLDBIG2.columns = ["region", "commodity", "year", "type", "unit", "value"]
WORLDBIG1_2017 = WORLDBIG1[WORLDBIG1["year"] == 2017]
WORLDBIG2_2017 = WORLDBIG2[WORLDBIG2["year"] == 2017]
WORLDBIG_2017 = pd.concat([WORLDBIG1_2017, WORLDBIG2_2017], ignore_index=True)
print(WORLDBIG_2017.head())
```

```python
#지역코드제거
exclude_regions = [
    "WORLD", "OECDAM", "OECDAO", "OECDEUR", "AFRICA", "LATAMER", "MIDEAST", "EURASIA", "ASIA", "WORLDMAR", "WORLDAV",
    "UNAFRICA", "UNAMERICAS", "UNASIATOT", "UNEUROPE", "UNOCEANIA", "OECDTOT", "NOECDTOT", "IEATOT", "IEAFAMILY",
    "EU28", "EU27_2020", "MFSU15", "MYUGO", "OPEC", "MASEAN", "MG7", "MG8", "MG20", "CHINAREG"
]
WORLDBIG_2017 = WORLDBIG_2017[~WORLDBIG_2017['region'].isin(exclude_regions)]
TOE = WORLDBIG_2017[WORLDBIG_2017['unit'] == 'KTOE'].copy()
TOE['value'] = pd.to_numeric(TOE['value'], errors='coerce').fillna(0)
print(TOE.head())
```

```python
#국가코드병합
country_code = pd.read_csv("country_code_250805.csv")
country_code.columns = ["region", "region_code"]
TOE = TOE.merge(country_code, on='region', how='left')
print(TOE.head())
```

```python
p_order_df = pd.read_excel("p_order.xlsx")
f_order_df = pd.read_excel("f_order.xlsx")
# ordering standard
p_order_vec = p_order_df.sort_values('products_order')['products'].tolist()
f_order_vec = f_order_df.sort_values('flow_order')['flow'].tolist()

# 고유 region_code 목록
region_codes = TOE['region_code'].dropna().unique()
print(len(region_codes))
p_order_df.head(), f_order_df.head()
```

```python
# 4. region_code별로 Excel 시트 생성
wb = Workbook()
wb.remove(wb.active)  # 기본 시트 제거
for region in region_codes:
    region_df = TOE[TOE['region_code'] == region]
    pivot_table = pd.pivot_table(
        region_df,
        index='type',
        columns='commodity',
        values='value',
        aggfunc='sum',
        fill_value=0
    ).reset_index() 
    #type ordering
    pivot_table['type'] = pd.Categorical(pivot_table['type'], categories=f_order_vec, ordered=True)
    pivot_table = pivot_table.sort_values('type')
    
    col_order = ['type'] + [p for p in p_order_vec if p in pivot_table.columns]
    pivot_table = pivot_table[col_order]

    # 워크시트 추가
    ws = wb.create_sheet(title=region[:10])
    for r in dataframe_to_rows(pivot_table, index=False, header=True):
        ws.append(r)
```

```python
wb.save("WEB_250805.xlsx")
```
