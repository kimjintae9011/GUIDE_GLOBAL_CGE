# Apply Function

```python
def R17(Country_Code):
    if Country_Code in R17_01KOR  :
        return '01_KOR'

    if Country_Code in R17_02CHN  :
        return '02_CHN'   

    if Country_Code in R17_03JPN  :
        return '03_JPN'   

    if Country_Code in R17_04RUS  :
        return '04_RUS'   

    if Country_Code in R17_05MNG  :
        return '05_MNG'     

    if Country_Code in R17_06PRK  :
        return '06_PRK' 
    
    if Country_Code in R17_07NAM  :
        return '07_NAM' 
    
    if Country_Code in R17_08LAM  :
        return '08_LAM'  

    if Country_Code in R17_09WEU  :
        return '09_WEU'  

    if Country_Code in R17_10EEU  :
        return '10_EEU'      

    if Country_Code in R17_11RFSU  :
        return '11_RFSU' 

    if Country_Code in R17_12MEA  :
        return '12_MEA' 
    
    if Country_Code in R17_13AFR  :
        return '13_AFR'

    if Country_Code in R17_14RCPA  :
        return '14_RCPA'

    if Country_Code in R17_15SAS  :
        return '15_SAS'
    
    if Country_Code in R17_16RPAS  :
        return '16_RPAS'

    if Country_Code in R17_17RPAO  :
        return '17_RPAO'
```

```python
df_gdp['R17'] = df_gdp['Country Name'].apply(R17)
df_pop['R17'] = df_pop['Country Name'].apply(R17)
```

```python
df_gdp.to_excel('test_gdp.xlsx')
df_pop.to_excel('test_pop.xlsx')
```

# Filtering

```python
Year = ['R17','2018','2019', '2020', '2021','2022']
```

```python
GDPPPP = df_gdp[Year]
POP = df_pop[Year]
```

```python
pivot_table_GDPPPP = pd.pivot_table(GDPPPP,index='R17', aggfunc=sum)
pivot_table_POP = pd.pivot_table(POP, index='R17', aggfunc=sum)
```

```python
pivot_table_GDPPPP.transpose().to_excel('./Output/GDPPPP.xlsx')
pivot_table_POP.transpose().to_excel('./Output/POP.xlsx')
```
