# R17 

 Region code list
- See message_data.tools.regions.
- The ISO 3166-1 alpha-3 codes are not defined in this file, but loaded from
   a copy of the ISO database, e.g. in pycountry.
- Among others, there are no assignments for:
- ATA Antarctica
- IOT British Indian Ocean Territory
- SGS South Georgia

* https://docs.messageix.org/projects/models/en/v2021.3.22/pkg-data/node.html

# Apply Function

```python
def R17(Country_Code):
    if Country_Code in R17_01KOR  :
        return '01_KOR'

    if Country_Code in R17_02CHN  :
        return '02_CHN'   

    if Country_Code in R17_03JPN  :
        return '03_JPN'   

    if Country_Code in R17_04RUS  :
        return '04_RUS'   

    if Country_Code in R17_05MNG  :
        return '05_MNG'     

    if Country_Code in R17_06PRK  :
        return '06_PRK' 
    
    if Country_Code in R17_07NAM  :
        return '07_NAM' 
    
    if Country_Code in R17_08LAM  :
        return '08_LAM'  

    if Country_Code in R17_09WEU  :
        return '09_WEU'  

    if Country_Code in R17_10EEU  :
        return '10_EEU'      

    if Country_Code in R17_11FSU  :
        return '11_FSU' 

    if Country_Code in R17_12MEA  :
        return '12_MEA' 
    
    if Country_Code in R17_13AFR  :
        return '13_AFR'

    if Country_Code in R17_14CPA  :
        return '14_CPA'

    if Country_Code in R17_15SAS  :
        return '15_SAS'
    
    if Country_Code in R17_16PAS  :
        return '16_PAS'

    if Country_Code in R17_17PAO  :
        return '17_PAO'
```

```python
df_GDP['R17'] = df_GDP['Country Code'].apply(R17)
df_POP['R17'] = df_POP['Country Code'].apply(R17)
df_Labor['R17'] = df_Labor['Country Code'].apply(R17)
df_Capital['R17'] = df_Capital['Country Code'].apply(R17)
df_Land['R17'] = df_Land['Country Code'].apply(R17)
```
